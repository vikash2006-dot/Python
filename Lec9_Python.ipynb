```python
class student:
  # constructor
  def __init__(self,name,rollno,age):
    self.name=name
    self.rollno=rollno
    self.age=age

  def display(self):
    print(self.name,self.rollno,self.age)


temp=student("Sachin",51,20)
print(temp.name)
print(temp.rollno)
print(temp.age)

temp.display()

# all bounded parameters in class is known as encapuslation
```
