```python
import numpy as np

# selecting elements of matrix
a=np.random.randint(1,90,25);

# Giving dim to matrix
b=a.reshape((5,5))
print (b)

# row wise mean of matrix
print(f"mean of all rows {b.mean(axis=1)}")

# column wise mean of matrix
print(f"mean of all columns {b.mean(axis=0)}")

# replacing all even numbers with -1
print(" replacing even number with -1 ............")
b=np.where(b%2==0,-1,b)

# row wise sorting matrix
print("row wise sorting matrix .............")
print(np.sort(b,axis=0))

# column wise sorting matrix
print("column wise sorting matrix ...............")
print(np.sort(b,axis=1))

# row wise sum of matrix
print("row wise sum of matrix ..................")
print(b.sum(axis=1))

# column wise sum of matrix
print("column wise sum of matrix ................")
print(b.sum(axis=0))

# counting -1 column wise
print("counting -1 column wise ................")
cnt=np.sum(b== -1,axis=0)
print(cnt)
# method 2
print("method 2........")
cnt1=np.sum(np.where(b<0,1,0),axis=0)
print(cnt1)
```

```python
# Question 2
import numpy as np
a=np.random.randint(1,90,25);
# or stock=np.random.randint(1,90,(5,5))
b=np.random.randint(1,90,25);
stock=a.reshape((5,5));
demand=b.reshape((5,5));
print("stocks:")
print(stock)
print("\n")
print("Demand:")
print(demand)
print("\n")

print("remaining stocks:")
rem_stock=stock-demand
print(rem_stock)
print("\n")

print("replacing negative stocks with 0..........")
s=np.where(rem_stock<0,0,rem_stock)
print(s)
print("\n")

print("shortage:")
shortage=np.sum(np.where(rem_stock<0,-1*rem_stock,0),axis=0)
print(shortage)
print("\n")

print("No shortage product:")
no_shortage=np.sum(np.where(rem_stock>=0,1,0))
print(no_shortage)
print("\n")

print("sorted remaining product:")
sorted_rem_product=np.sort(rem_stock,axis=1)
print(sorted_rem_product)
```
