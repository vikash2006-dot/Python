```python
lst=[1,2,3]
# append
# append fuction takes only one argument and append as it is to the last of the lst
lst.append([2,3])
print (lst)

#extend
#extend function takes only one argument and append every value as an individual value in lst

lst.extend([5,5,6])
print(lst)

#insert(index,value)
#insert element at a particular index
lst.insert(10,20)
print(lst)

#pop()
#delete last element of the list
lst.pop()
print (lst)

#remove()
#delete the first occurance of the element
lst.remove(3)
print(lst)
```

```python
lst=["hello",True,76,'A']

# index
#index function returns the index of the element
print(lst.index(76))
lst.append(76)
lst.extend({2,3,False})

# count()
# count function counts the occurance of the element in the list
print(lst.count(76))

# sort()
lst2=[1,5,7,3,7]
lst2.sort()
print(lst2)

# min , max , sum
print(min(lst2))
print(sum(lst2))

#sorted
#sorted function returns a new sorted list
print(sorted(lst2))
```
